### Overall functionality requirements

```python
import numpy as cp
import sys
cp.set_printoptions(threshold=sys.maxsize)
```

```python
bitwidth = 8
prob = 0.25
```

## LLN Noise Convergence

```python
n_noise = 0
epoch = 1000
```

```python
for _ in range(epoch):
    noise_mask_arr = cp.random.rand(1, bitwidth)
    noise_mask_arr = (noise_mask_arr < prob).astype(cp.uint32)
    n_noise += int(cp.sum(noise_mask_arr))
    
percent_noisy = n_noise / (bitwidth * epoch)
print(percent_noisy)
```

```python
noise_mask = cp.sum(noise_mask_arr * (2 ** cp.arange(bitwidth-1,-1,-1)), axis=1)
noise_mask = noise_mask.astype(cp.uint32)
print(2 ** (cp.arange(bitwidth-1,-1,-1)))
```
